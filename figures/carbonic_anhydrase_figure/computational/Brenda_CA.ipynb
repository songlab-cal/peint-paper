```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib as mpl

import seaborn as sns
mpl.rcParams['pdf.fonttype'] = 42
mpl.rcParams['ps.fonttype'] = 42
```

```python
datafile = 'barr_evan_params.csv'
df = pd.read_csv(datafile)

ca_subset = df[(df['EC1'] == 4) &
               (df['EC2'] == 2) &
               (df['EC3'] == 1) &
               (df['EC4'] == 1)]

print(len(ca_subset))
```

```python
fig, axs = plt.subplots(figsize=(0.7,3))

sns.boxplot(data=ca_subset[ca_subset['kcat (1/sec)'] > 1], y='kcat (1/sec)',notch=False, showcaps=True,
    flierprops={"marker": ""},
    boxprops={"facecolor": (.3, .5, .7, .5)},
    medianprops={"color": "black", "linewidth": 0.5},
    linewidth=0.5)
sns.stripplot(data = ca_subset[ca_subset['kcat (1/sec)'] > 1], y='kcat (1/sec)', color='green', s=5, linewidth=0.2)

#Prior D152 mutations from literature
axs.axhline(y=0.4e4, color='red', linestyle='--', linewidth=0.5)
axs.axhline(y=1e4, color='red', linestyle='--', linewidth=0.5)

for spine in axs.spines.values():
    spine.set_linewidth(0.5)

axs.set_ylim(100, 1.5e6)
axs.tick_params(axis='both', which='both', width=0.5)
axs.set_yscale('log')
axs.set_xticks([])
plt.savefig('Brenda_CA.pdf', bbox_inches='tight')
```
